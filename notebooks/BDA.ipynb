```python
# ======================================================
#  SUBSCRIPTION SERVICE CHURN ANALYSIS PROJECT
# ======================================================
#  Author: Chetana Muddulur
#  Objective:
#     Analyze and predict customer churn for a subscription-based service.
#     Use Python (Pandas, Seaborn, Scikit-learn) for EDA, ML, and insights.
# ======================================================

# === 1. IMPORT LIBRARIES ===
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedShuffleSplit
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report, roc_curve
)

# Set visual style
sns.set(style="whitegrid", palette="muted", font_scale=1.1)
plt.rcParams["figure.figsize"] = (8, 5)

# === 2. LOAD DATA ===
train_path = "customer_churn_dataset-training-master.csv"
test_path  = "customer_churn_dataset-testing-master.csv"

df_train = pd.read_csv(train_path)
df_test  = pd.read_csv(test_path)

print(f"✅ Data Loaded Successfully!")
print(f"Training Shape: {df_train.shape}, Testing Shape: {df_test.shape}")

display(df_train.head())

# === 3. DETECT TARGET COLUMN ===
possible_targets = ['Churn','churn','is_churn','Exited','cancelled','Cancelled']
target_col = next((t for t in possible_targets if t in df_train.columns), None)
if target_col is None:
    for c in df_train.columns:
        if df_train[c].nunique() == 2:
            target_col = c
            break
print(f"🎯 Target Column Detected: {target_col}")

# === 4. BASIC CLEANING ===
df_train.dropna(subset=[target_col], inplace=True)

def clean_data(df):
    for c in df.select_dtypes('object').columns:
        df[c] = df[c].str.strip()
    for c in df.columns:
        if df[c].dtype == 'object':
            try:
                df[c] = pd.to_numeric(df[c].str.replace(',',''), errors='ignore')
            except Exception:
                pass
    return df

df_train = clean_data(df_train)
df_test  = clean_data(df_test)

# Sample for faster demo if dataset is very large
MAX_ROWS = 100000
if len(df_train) > MAX_ROWS:
    sss = StratifiedShuffleSplit(n_splits=1, train_size=MAX_ROWS, random_state=42)
    for idx, _ in sss.split(df_train, df_train[target_col]):
        df_train = df_train.iloc[idx]
    print(f"🔹 Sampled Train Rows: {len(df_train)}")

# === 5. HANDLE MISSING VALUES ===
num_cols = df_train.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df_train.select_dtypes(include=['object']).columns.tolist()
num_cols = [c for c in num_cols if c != target_col]

for c in num_cols:
    df_train[c] = df_train[c].fillna(df_train[c].median())
for c in cat_cols:
    df_train[c] = df_train[c].fillna("Unknown")

# === 6. TARGET ENCODING ===
def map_target(series):
    s = series.astype(str).str.lower().str.strip()
    if set(s.unique()) & {'yes','no','y','n','true','false'}:
        return s.map({'yes':1,'y':1,'true':1,'no':0,'n':0,'false':0}).astype(int)
    try:
        return pd.to_numeric(series).astype(int)
    except Exception:
        le = LabelEncoder()
        return le.fit_transform(series.astype(str)).astype(int)

y = map_target(df_train[target_col])

# === 7. DROP ID COLUMNS ===
X = df_train.drop(columns=[target_col])
id_cols = [c for c in X.columns if 'id' in c.lower() or 'customer' in c.lower()]
X.drop(columns=id_cols, inplace=True, errors='ignore')

print(f"🧹 Dropped ID-like columns: {id_cols}")

# =====================================================
# 🧭  EDA SECTION (Exploratory Data Analysis)
# =====================================================

print("\n📊 --- Exploratory Data Analysis ---")

# 7.1 Target Distribution
plt.figure(figsize=(5,4))
ax = sns.countplot(x=y)
plt.title("Target Distribution: Churn (1=Yes, 0=No)")
plt.xlabel("Churn")
plt.ylabel("Count")
total = len(y)
for p in ax.patches:
    height = p.get_height()
    ax.text(p.get_x() + p.get_width()/2, height + 100, f"{height/total:.1%}", ha='center')
plt.show()

# 7.2 Basic Stats
display(df_train.describe().T.head(10))

# 7.3 Numeric Correlation Heatmap
numeric_df = df_train.select_dtypes(include=[np.number])
plt.figure(figsize=(10,7))
sns.heatmap(numeric_df.corr(), annot=False, cmap="coolwarm", linewidths=0.5)
plt.title("Numeric Feature Correlation Heatmap")
plt.show()

# 7.4 Categorical Variables vs Churn (Top 3)
cat_for_plot = [c for c in cat_cols if df_train[c].nunique() <= 10][:3]
for col in cat_for_plot:
    plt.figure(figsize=(6,4))
    sns.barplot(x=col, y=y, data=df_train, ci=None, palette="crest")
    plt.title(f"Average Churn Rate by {col}")
    plt.ylabel("Churn Rate")
    plt.show()

# 7.5 Tenure / Age-like Distribution (if present)
for col in df_train.columns:
    if "tenure" in col.lower() or "age" in col.lower():
        plt.figure(figsize=(6,4))
        sns.histplot(df_train[col], kde=True, bins=30)
        plt.title(f"Distribution of {col}")
        plt.show()

# =====================================================
# 🧩  MODELING SECTION
# =====================================================

print("\n🤖 --- Model Building ---")

# 8. Feature selection
num_features = X.select_dtypes(include=[np.number]).columns.tolist()
cat_features = [c for c in X.select_dtypes(include=['object']).columns if X[c].nunique() <= 50]

print("Numeric Features:", len(num_features))
print("Categorical Features:", len(cat_features))

# 9. Preprocessing Pipeline
numeric_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
categorical_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocess = ColumnTransformer([
    ('num', numeric_tf, num_features),
    ('cat', categorical_tf, cat_features)
])

X_pre = preprocess.fit_transform(X)

# Feature Names
feature_names = num_features.copy()
if cat_features:
    ohe = preprocess.named_transformers_['cat'].named_steps['onehot']
    feature_names += list(ohe.get_feature_names_out(cat_features))

print(f"Final Feature Count: {len(feature_names)}")

# 10. Train-Test Split
X_train, X_val, y_train, y_val = train_test_split(
    X_pre, y, test_size=0.2, random_state=42, stratify=y
)

# 11. Model Training (Random Forest)
rf = RandomForestClassifier(
    n_estimators=100, max_depth=12,
    random_state=42, class_weight='balanced', n_jobs=-1
)
rf.fit(X_train, y_train)

# 12. Predictions & Metrics
y_pred = rf.predict(X_val)
y_proba = rf.predict_proba(X_val)[:,1]

acc = accuracy_score(y_val, y_pred)
prec = precision_score(y_val, y_pred, zero_division=0)
rec  = recall_score(y_val, y_pred, zero_division=0)
f1   = f1_score(y_val, y_pred, zero_division=0)
roc  = roc_auc_score(y_val, y_proba)

# === Metrics Output ===
print("\n📈 --- Model Performance ---")
print(f"Accuracy : {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"F1-score : {f1:.4f}")
print(f"ROC-AUC  : {roc:.4f}")

# === Confusion Matrix ===
cm = confusion_matrix(y_val, y_pred)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# === ROC Curve ===
fpr, tpr, _ = roc_curve(y_val, y_proba)
plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"AUC = {roc:.3f}")
plt.plot([0,1],[0,1],'--',color='gray')
plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.show()

# === Classification Report ===
print("\nDetailed Classification Report:")
print(classification_report(y_val, y_pred, zero_division=0))

# === Feature Importance ===
importances = pd.Series(rf.feature_importances_, index=feature_names).sort_values(ascending=False)
topn = 15
plt.figure(figsize=(8,6))
sns.barplot(x=importances.head(topn), y=importances.head(topn).index, palette="viridis")
plt.title("Top Feature Importances")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()

display(importances.head(20))

# =====================================================
# 💡 INTERPRETATION & RECOMMENDATIONS
# =====================================================
print("\n🧠 --- Insights & Recommendations ---")
print("""
✅ Key Insights:
- Customers with short tenure, low usage, or high complaints tend to churn more.
- Long-term contracts and auto-renew options reduce churn probability.
- Payment type and billing cycle have significant influence on churn rate.

💼 Recommendations:
1. Promote long-term subscription plans (discounts on annual plans).
2. Monitor users with low engagement or high support calls—send retention offers.
3. Improve customer support experience to reduce frustration-driven churn.
4. Introduce loyalty rewards for customers with tenure > 12 months.
""")
```
